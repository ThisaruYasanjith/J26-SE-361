# 29F — Qwen Explanation Quality Validation

Comparison of existing Qwen2.5-Coder-7B-Instruct
base-model and fine-tuned LoRA explanation evidence.

Recorded generation times and text examples are experimental
observations. They do **not** establish human-rated quality.

Quality dimensions for future blinded review:
correctness, relevance, actionability, grounding,
and unsupported claims.

The sample is a small controlled demonstration, not
representative external evaluation.


```python
from pathlib import Path
import pandas as pd
from IPython.display import display

root = Path(r"C:\4th Year\Research Work\Implementation Stage\J26-SE-361")
folder = root / "Code Cleanliness" / "evaluation" / "results" / "final_evaluation" / "llm"

df = pd.read_csv(folder / "base_vs_finetuned_review.csv")

print("LLM comparison records:", len(df))
print("Columns:", list(df.columns))
display(df.head(10))
```

```python
# Review-completeness audit.
review_columns = [
    c for c in df.columns
    if any(w in c.lower() for w in
           ["rating", "score", "reviewer", "human",
            "correctness", "relevance", "grounding",
            "actionability", "unsupported"])
]

print("Potential quality review columns:")
for c in review_columns:
    print(c, "non-empty:", df[c].notna().sum())

print("Do not interpret unreviewed explanations as validated quality improvements.")
```
